```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('laptops.csv')
```

```python
df.head()
```

```python
df.describe()
```

```python
print("\nData types and unique values:")
print(df.dtypes)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
df = df.fillna(df.mean(numeric_only=True))
```

```python
print("Missing values now:")
print(df.isnull().sum())
```

```python
df['Screen_Size_cm'] = df['Screen_Size_cm'].round(2)
```

```python
df['Weight_kg'] = df['Weight_kg'].round(2)
```

```python
df.head()
```

```python
df['Screen_Size_in'] = df['Screen_Size_cm'] * 0.39
```

```python
df['Weight_lb'] = df['Weight_kg'] * 2.2
```

```python
df[['Screen_Size_cm', 'Screen_Size_in', 'Weight_kg', 'Weight_lb']].head()
```

```python
df['Screen_Size_in'] = df['Screen_Size_in'].round(2)
```

```python
df['Weight_lb'] = df['Weight_lb'].round(2)
```

```python
df[['Screen_Size_cm', 'Screen_Size_in', 'Weight_kg', 'Weight_lb']].head()
```

```python
df['Price-binned'] = pd.qcut(df['Price'], q=3, labels=['Low', 'Medium', 'High'])
print("\nPrice bin distribution:\n", df['Price-binned'].value_counts())
```

```python
df[['Price', 'Price-binned']].head(10)
```

```python
df[['Screen']].head(10)
```

```python
print(df['Screen'].nunique())
```

```python
def screen_encoder(value):
    if value == "IPS Panel":
        return 1
    elif value == "Full HD":
        return 0
    else:
        return -1
```

```python
df['Screen_encoded'] = df['Screen'].apply(screen_encoder)
```

```python
df[['Screen', 'Screen_encoded']].head()
```

```python
df.to_csv('laptops_processed.csv', index=False)
print("\nProcessed dataset saved as 'laptops_processed.csv'")
```
